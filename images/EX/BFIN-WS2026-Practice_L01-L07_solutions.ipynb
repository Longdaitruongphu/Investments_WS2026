### Question 1 (Easy)

Alpha purchases a newly built production line from Beta. Instead of paying cash, Alpha issues new bonds directly to Beta in payment. A week later, Beta sells those bonds to Gamma for cash. Alpha continues to own and use the production line.

a. For each transaction, identify the real asset and financial claims involved. State who holds the bonds and who owes the bond payments after each transaction.

b. Beta describes the second transaction as “another round of financing for Alpha.” Assess this description.

---


### Solution to Question 1

#### Detailed Solution

**Part a. Real asset vs. financial claims & ownership structure:**
- **Real Asset:** The *newly built production line* (a physical productive capital asset capable of generating goods, services, and future cash flows).
- **Financial Claims:** The *newly issued bonds* (debt contracts representing promises by the issuer to pay future cash flows).
- **After Transaction 1 (Alpha issues bonds to Beta for the production line):**
  - **Real Asset Owner:** Alpha owns and utilizes the production line.
  - **Bondholder (Creditor):** Beta holds the bonds.
  - **Obligor (Debtor):** Alpha owes the debt service (coupons and principal payments) to Beta.
- **After Transaction 2 (Beta sells Alpha's bonds to Gamma for cash):**
  - **Real Asset Owner:** Alpha continues to own and utilize the production line.
  - **Bondholder (Creditor):** Gamma now holds the bonds.
  - **Obligor (Debtor):** Alpha now owes the debt service to Gamma. Beta holds cash and has no further claims or liabilities related to Alpha.

**Part b. Assessment of Beta's statement:**
- **Assessment:** Beta's description is **incorrect**.
- **Explanation:** Transaction 1 was a *primary market* transaction that provided financing for Alpha (allowing Alpha to acquire a real asset by issuing new securities). In contrast, Transaction 2 is a *secondary market* transaction between Beta and Gamma. Alpha was not a party to this trade, issued no new claims, and received no new capital. The secondary market merely transferred ownership of existing financial claims from one investor (Beta) to another (Gamma), providing liquidity to Beta without altering Alpha's capital base.

---
#### Financial Interpretation
Real assets produce the productive output of an economy, whereas financial claims determine the allocation of income and wealth among investors. Primary markets channel capital directly to issuers to fund real investments, while secondary markets provide the essential liquidity and price discovery that make investors willing to purchase primary securities in the first place.


### Question 2 (Easy)

Delta has issued ordinary bonds, preferred shares and common shares. An investor makes these statements:

| Statement | Claim |
|---|---|
| A | Preferred shareholders rank ahead of common shareholders, so they must also rank ahead of bondholders. |
| B | A stated preferred dividend is a contractual debt payment; omitting it has the same consequence as failing to pay a bond coupon. |
| C | Common shareholders have limited liability, so they cannot lose the money they paid for their shares. |

Assess each statement and give a corrected explanation. Assume ordinary securities with no special contractual provisions; detailed insolvency rules are not required.

---


### Solution to Question 2

#### Detailed Solution

**Statement A:** *"Preferred shareholders rank ahead of common shareholders, so they must also rank ahead of bondholders."*
- **Assessment:** **False**.
- **Correction:** In the priority of claims (the corporate capital structure), debt holders (bondholders) have the highest priority. Preferred stock is an equity instrument: preferred shareholders rank ahead of common equity, but they are strictly **subordinate to all debt holders**. In liquidation or distress, bondholders must be paid in full before preferred shareholders receive any residual distribution. Priority order: **Senior Debt $\rightarrow$ Subordinated Debt $\rightarrow$ Preferred Stock $\rightarrow$ Common Stock**.

**Statement B:** *"A stated preferred dividend is a contractual debt payment; omitting it has the same consequence as failing to pay a bond coupon."*
- **Assessment:** **False**.
- **Correction:** Bond coupons are legal, binding debt obligations; failure to pay constitutes a **legal default** that can force the firm into bankruptcy or insolvency proceedings. In contrast, preferred dividends are declared at the discretion of the Board of Directors. Omitting a preferred dividend does *not* trigger default or bankruptcy. Typically, preferred stock is *cumulative*, meaning omitted dividends accumulate in arrears and must be paid in full before common shareholders can receive any dividends, but nonpayment does not precipitate bankruptcy.

**Statement C:** *"Common shareholders have limited liability, so they cannot lose the money they paid for their shares."*
- **Assessment:** **False**.
- **Correction:** Limited liability means that shareholders are not personally liable for the firm's debts beyond their investment; their personal assets are protected. However, shareholders **can lose up to 100% of the capital invested** (the purchase price paid for the shares) if the firm fails and its equity value drops to zero. Their loss is capped at their investment, not zero.

---
#### Financial Interpretation
Understanding the absolute priority rule and the structural differences between debt, mezzanine/preferred equity, and residual common equity is essential for risk assessment. Debt provides fixed, legally enforceable claims with low risk, preferred shares provide higher yields with subordinated discretionary cash flows, and common equity bears the residual operational and financial risk with unlimited upside but total loss exposure.


### Question 3 (Easy)

Two managers describe their investment decisions:

| Manager | Decision |
|---|---|
| Epsilon | After reviewing inflation and economic growth forecasts, reduces the portfolio's equity allocation from 60% to 40%. Epsilon then chooses individual bonds by comparing issuers' credit quality and yields. |
| Zeta | Finds Eta's shares attractively priced after studying the firm's cash flows and competitive position. Zeta buys the shares despite an unfavourable industry outlook. |

a. Classify each manager's investment process and identify the asset-allocation or security-selection decisions.

b. Epsilon claims the strategy is passive because the portfolio holds many securities. Is that sufficient to support the claim?

---


### Solution to Question 3

#### Detailed Solution

**Part a. Classification of investment processes and decisions:**
- **Manager Epsilon:**
  - **Investment Process:** **Top-Down Investing**.
  - **Asset Allocation Decision:** Shifting the macro portfolio allocation from 60% equity to 40% equity based on macroeconomic forecasts (inflation and growth).
  - **Security Selection Decision:** Evaluating individual bond issuers' credit quality and yields to choose specific bonds within the fixed income allocation.
- **Manager Zeta:**
  - **Investment Process:** **Bottom-Up Investing**.
  - **Asset Allocation Decision:** Not actively managed at the macro level (the macro/industry outlook is ignored).
  - **Security Selection Decision:** Conducting fundamental firm-level research (cash flows, competitive position) to identify an underpriced stock (Eta) regardless of the unfavorable industry environment.

**Part b. Assessment of Epsilon's claim of a passive strategy:**
- **Assessment:** The claim is **unjustified / false**.
- **Explanation:** A passive investment strategy aims to replicate the return of a broad market index or benchmark without attempting to outguess macroeconomic trends or mispriced securities. Epsilon is actively making **tactical asset allocation** bets by timing the market (reducing equity from 60% to 40% based on economic forecasts) and practicing **active security selection** by evaluating credit quality and spreads. Holding a large number of securities achieves **diversification**, but diversification is distinct from passivity. A diversified portfolio can be actively managed.

---
#### Financial Interpretation
Top-down managers prioritize macro trends and asset allocation (which historically explains over 85–90% of portfolio return variance), whereas bottom-up managers focus on micro-level mispricings. Passive management requires adherence to rule-based index tracking; active management involves discretionary timing and selection.


### Question 4 (Average)

Four investors take separate positions on Theta shares. All contracts expire on the same date.

| Investor | Position | Strike or delivery price per share |
|---|---|---:|
| Iota | Buys a put option | USD 50 |
| Kappa | Enters a short forward | USD 50 |
| Lambda | Buys a call option | USD 50 |
| Mu | Enters a long forward | USD 50 |

Option premiums are paid when the positions are opened. Assume no counterparty default.

a. At expiration, Theta trades at USD 62. Should Iota exercise the put? Can Kappa choose not to deliver?

b. Consider the alternative outcome in which Theta instead trades at USD 38. Should Lambda exercise the call? Can Mu decline to buy?

c. An adviser says, “An option buyer can walk away, so an option that is not exercised costs the buyer nothing.” Assess the statement. No profit calculations are required.

---


### Solution to Question 4

#### Detailed Solution

**Positions:**
- Iota: Long Put ($K = 50$)
- Kappa: Short Forward ($F_0 = 50$)
- Lambda: Long Call ($K = 50$)
- Mu: Long Forward ($F_0 = 50$)

**Part a. Outcome: Terminal share price $S_T = \text{USD } 62$:**
- **Iota (Long Put):** Strike $K = 50$, Market $S_T = 62$. The put option is *out of the money* ($S_T > K$). Exercising would yield a payoff of $\max(0, 50 - 62) = 0$. Selling the share in the open market yields \$62, which is better than selling at the strike of \$50. Therefore, Iota **should NOT exercise** the put.
- **Kappa (Short Forward):** A forward contract represents a **binding legal commitment**, not an option. Kappa **CANNOT choose not to deliver**; Kappa is legally obligated to deliver the shares (or cash settle) at \$50, incurring a loss of $50 - 62 = -\text{USD } 12$.

**Part b. Outcome: Terminal share price $S_T = \text{USD } 38$:**
- **Lambda (Long Call):** Strike $K = 50$, Market $S_T = 38$. The call option is *out of the money* ($S_T < K$). Exercising would mean buying at \$50 when the market price is \$38, yielding $\max(0, 38 - 50) = 0$. Therefore, Lambda **should NOT exercise** the call.
- **Mu (Long Forward):** A forward contract is a bilateral obligation. Mu **CANNOT decline to buy**; Mu is legally obligated to purchase the shares at \$50, incurring a loss of $38 - 50 = -\text{USD } 12$.

**Part c. Assessment of adviser's statement:**
- **Assessment:** The statement is **misleading / incorrect**.
- **Explanation:** While an option buyer has the right to walk away from exercising at expiration (avoiding negative terminal payoffs), purchasing an option requires paying an upfront **option premium** at inception. If the option expires unexercised, the buyer loses the **entire initial premium paid**. Thus, an unexercised option costs the buyer the full initial capital outlay.

---
#### Financial Interpretation
Options provide asymmetric payoff profiles (conferring rights without obligations in exchange for an upfront sunk premium), whereas forward and futures contracts provide symmetric payoff profiles with zero initial cost but unconditional binding commitments for both counterparties.


### Question 5 (Average)

Nu sells Treasury securities to Xi for USD 240,000 today and agrees to repurchase the same securities from Xi for USD 241,080 after 45 days. No coupon is paid during the agreement, no fees apply, and both parties perform as agreed.

a. Identify the cash borrower and lender. Which party describes its position as a repo, and which as a reverse repo?

b. Show the cash flows for each party today and on day 45. Calculate the lender's financing income.

c. Calculate the lender's 45-day return and its simple annualized rate on a 360-day basis.

---


### Solution to Question 5

#### Detailed Solution

**Part a. Roles and descriptions:**
- **Cash Borrower:** **Nu** (sells collateral securities today and repurchases them later at a higher price).
- **Cash Lender:** **Xi** (provides cash today and receives cash plus financing interest later).
- **Terminology:**
  - Nu describes the transaction as a **Repo (Repurchase Agreement)**.
  - Xi describes the transaction as a **Reverse Repo (Reverse Repurchase Agreement)**.

**Part b. Cash flows and financing income:**
- **Today ($t = 0$):**
  - Nu receives: **+\$240,000**
  - Xi pays: **-\$240,000**
- **Day 45 ($t = 45$):**
  - Nu pays: **-\$241,080**
  - Xi receives: **+\$241,080**
- **Lender's (Xi's) Financing Income:**
  $$\text{Income} = \text{Repurchase Price} - \text{Sale Price} = \$241,080 - \$240,000 = \mathbf{\$1,080}$$

**Part c. 45-day return and annualized rate (360-day basis):**
- **45-day Holding Period Return (HPR):**
  $$\text{HPR} = \frac{\$1,080}{\$240,000} = 0.0045 = \mathbf{0.45\%}$$
- **Simple Annualized Repo Rate ($r_{\text{repo}}$) on a money market 360-day basis:**
  $$r_{\text{repo}} = \text{HPR} \times \left(\frac{360}{\text{Days}}\right) = 0.0045 \times \left(\frac{360}{45}\right) = 0.0045 \times 8 = 0.0360 = \mathbf{3.60\%}$$

---
#### Financial Interpretation
Repos are vital short-term secured money market instruments. By using high-quality government securities as collateral, institutions borrow short-term liquidity at low financing rates (3.60% annualized), while cash lenders earn secure, collateralized interest.


### Question 6 (Average)

Two options on Omicron shares expire in one month:

| Purchased option | Strike price per share | Premium per share |
|---|---:|---:|
| Call | USD 46 | USD 3.50 |
| Put | USD 50 | USD 2.25 |

Consider each purchase separately. The options can be exercised only at expiration.

a. For terminal share prices of USD 42, USD 48 and USD 56, calculate each option's payoff and its buyer's net profit per share.

b. Find each buyer's break-even terminal share price.

c. At USD 48, what should each option holder do? Explain.

---


### Solution to Question 6

#### Detailed Solution

**Specifications:**
- Long Call: Strike $K_C = \text{USD } 46$, Premium $C_0 = \text{USD } 3.50$.
- Long Put: Strike $K_P = \text{USD } 50$, Premium $P_0 = \text{USD } 2.25$.

**Part a. Payoffs and net profits per share:**
- Call formulas: $\text{Payoff} = \max(0, S_T - 46)$, $\text{Profit} = \text{Payoff} - 3.50$.
- Put formulas: $\text{Payoff} = \max(0, 50 - S_T)$, $\text{Profit} = \text{Payoff} - 2.25$.

| Terminal Price ($S_T$) | Call Payoff | Call Net Profit | Put Payoff | Put Net Profit |
|:---:|:---:|:---:|:---:|:---:|
| **USD 42** | $\max(0, 42-46) = \mathbf{0.00}$ | $0 - 3.50 = \mathbf{-\$3.50}$ | $\max(0, 50-42) = \mathbf{\$8.00}$ | $8.00 - 2.25 = \mathbf{+\$5.75}$ |
| **USD 48** | $\max(0, 48-46) = \mathbf{\$2.00}$ | $2.00 - 3.50 = \mathbf{-\$1.50}$ | $\max(0, 50-48) = \mathbf{\$2.00}$ | $2.00 - 2.25 = \mathbf{-\$0.25}$ |
| **USD 56** | $\max(0, 56-46) = \mathbf{\$10.00}$ | $10.00 - 3.50 = \mathbf{+\$6.50}$ | $\max(0, 50-56) = \mathbf{0.00}$ | $0 - 2.25 = \mathbf{-\$2.25}$ |

**Part b. Break-even terminal share prices:**
- **Call Buyer Break-Even:**
  $$S_{T, \text{break-even}}^{\text{Call}} = K_C + C_0 = 46 + 3.50 = \mathbf{\text{USD } 49.50}$$
- **Put Buyer Break-Even:**
  $$S_{T, \text{break-even}}^{\text{Put}} = K_P - P_0 = 50 - 2.25 = \mathbf{\text{USD } 47.75}$$

**Part c. Action at $S_T = \text{USD } 48$:**
- **Call Holder:** Should **EXERCISE**. The call is in the money ($S_T = 48 > K_C = 46$), delivering a positive payoff of \$2.00. While the net profit is negative ($-\text{\$1.50}$), the initial \$3.50 premium is a sunk cost. Exercising recoups \$2.00, reducing the net loss from \$3.50 to \$1.50.
- **Put Holder:** Should **EXERCISE**. The put is in the money ($S_T = 48 < K_P = 50$), delivering a positive payoff of \$2.00. Exercising recoups \$2.00, reducing the net loss from \$2.25 to \$0.25.

---
#### Financial Interpretation
Rational option exercise decisions at expiration depend strictly on **intrinsic value (payoff $\mathbf{> 0}$)**, completely ignoring the sunk purchase premium. The break-even price establishes the threshold above/below which net profitability turns positive.


### Question 7 (Average)

Pi bought a call option on Rho shares. The strike price and premium are missing from the record. A payoff worksheet shows the following possible expiration outcomes for that same option:

| Terminal share price (USD) | Buyer's net profit per share (USD) |
|---:|---:|
| 47 | −4 |
| 59 | +3 |
| 68 | +12 |

The worksheet includes the premium but no other costs.

a. Recover the premium and strike price.

b. Check your answer against all three rows. Calculate the payoff and net profit at a terminal share price of USD 54, and state whether Pi should exercise.

---


### Solution to Question 7

#### Detailed Solution

**Part a. Recover the premium and strike price:**
- The net profit for a long call option is given by:
  $$\text{Profit} = \max(0, S_T - K) - \text{Premium}$$
- When the call option is in the money ($S_T > K$), the payoff increases dollar-for-dollar with $S_T$:
  $$\frac{\Delta \text{Profit}}{\Delta S_T} = 1$$
- Look at the last two observations:
  $$\Delta S_T = 68 - 59 = 9, \quad \Delta \text{Profit} = 12 - 3 = 9$$
  Since $\Delta \text{Profit} / \Delta S_T = 9 / 9 = 1$, the option is strictly **in the money** at both $S_T = 59$ and $S_T = 68$.
- In contrast, at $S_T = 47$, the net profit is $-\$4$. If the option is out of the money at $S_T = 47$, the payoff is 0, so:
  $$\text{Profit} = 0 - \text{Premium} = -4 \implies \mathbf{\text{Premium} = \text{USD } 4.00}$$
- Now use the in-the-money profit equation at $S_T = 59$:
  $$\text{Profit} = (S_T - K) - \text{Premium} \implies 3 = (59 - K) - 4 \implies 59 - K = 7 \implies \mathbf{K = \text{USD } 52.00}$$

**Part b. Verification and analysis at $S_T = \text{USD } 54$:**
- **Verification across worksheet rows:**
  - At $S_T = 47$: $\max(0, 47 - 52) - 4 = 0 - 4 = -4$. *(Matches)*
  - At $S_T = 59$: $\max(0, 59 - 52) - 4 = 7 - 4 = +3$. *(Matches)*
  - At $S_T = 68$: $\max(0, 68 - 52) - 4 = 16 - 4 = +12$. *(Matches)*
- **At $S_T = \text{USD } 54$:**
  - **Payoff:** $\max(0, 54 - 52) = \mathbf{\text{USD } 2.00}$
  - **Net Profit:** $2.00 - 4.00 = \mathbf{-\text{USD } 2.00}$
  - **Exercise Decision:** Pi **SHOULD EXERCISE** because the option has positive intrinsic value ($S_T = 54 > K = 52$). Exercising recovers \$2.00 of intrinsic value, cutting the net loss in half compared to letting the option expire unexercised (which would lose the full \$4.00 premium).

---
#### Financial Interpretation
The piecewise linear kink in an option's payoff profile allows econometricians and traders to identify the strike price from the point where sensitivity to the underlying price shifts from 0 to 1.


### Question 8 (Hard)

Sigma has USD 10,000 to invest for one year. Tau shares currently cost USD 50 and pay no dividends. Sigma compares two fixed strategies:

| Strategy | Investment today |
|---|---|
| Shares | Buy 200 Tau shares. |
| Calls and cash | Buy 10 call contracts, each covering 100 Tau shares, at a premium of USD 5 per share. Invest the remaining money at a guaranteed annual return of 4%. |

The calls have a strike price of USD 55 and expire in one year. The contracts are cash-settled at expiration. No borrowing is allowed.

Sigma uses the following forecasts for Tau's price in one year:

| State | Probability | Terminal share price (USD) |
|---|---:|---:|
| Weak | 0.20 | 35 |
| Moderate | 0.40 | 55 |
| Strong | 0.40 | 75 |

a. Verify each strategy's initial cost and calculate the cash balance at the end of the year.

b. Calculate each strategy's terminal wealth and one-year return in every state.

c. Calculate each strategy's expected return and expected terminal wealth.

d. Sigma plans to select whichever strategy has the higher expected return. Assess this decision using your results.

---


### Solution to Question 8

#### Detailed Solution

**Part a. Verify initial costs and calculate year-end cash balance:**
- **Strategy 1 (Shares):**
  - Initial cost: $200 \text{ shares} \times \$50/\text{share} = \mathbf{\$10,000}$. Entire capital invested.
  - Ending cash balance: **\$0.00**.
- **Strategy 2 (Calls and Cash):**
  - 10 call contracts $\times 100 \text{ shares/contract} = 1,000 \text{ shares covered}$.
  - Call premium cost: $1,000 \text{ shares} \times \$5/\text{share} = \$5,000$.
  - Remaining cash: $\$10,000 - \$5,000 = \$5,000$.
  - Total initial outlay: $\$5,000 + \$5,000 = \mathbf{\$10,000}$.
  - Cash invested at guaranteed 4%:
    $$\text{Ending cash balance} = \$5,000 \times (1 + 0.04) = \mathbf{\$5,200}$$

**Part b. Terminal wealth and returns across states:**
- **State 1: Weak ($S_T = 35$, prob = 0.20):**
  - *Shares:* $W_T = 200 \times 35 = \mathbf{\$7,000}$; Return = $\frac{7,000 - 10,000}{10,000} = \mathbf{-30.0\%}$.
  - *Calls + Cash:* Call payoff = $1,000 \times \max(0, 35 - 55) = \$0$. $W_T = 0 + 5,200 = \mathbf{\$5,200}$; Return = $\frac{5,200 - 10,000}{10,000} = \mathbf{-48.0\%}$.
- **State 2: Moderate ($S_T = 55$, prob = 0.40):**
  - *Shares:* $W_T = 200 \times 55 = \mathbf{\$11,000}$; Return = $\frac{11,000 - 10,000}{10,000} = \mathbf{+10.0\%}$.
  - *Calls + Cash:* Call payoff = $1,000 \times \max(0, 55 - 55) = \$0$. $W_T = 0 + 5,200 = \mathbf{\$5,200}$; Return = $\mathbf{-48.0\%}$.
- **State 3: Strong ($S_T = 75$, prob = 0.40):**
  - *Shares:* $W_T = 200 \times 75 = \mathbf{\$15,000}$; Return = $\frac{15,000 - 10,000}{10,000} = \mathbf{+50.0\%}$.
  - *Calls + Cash:* Call payoff = $1,000 \times (75 - 55) = \$20,000$. $W_T = 20,000 + 5,200 = \mathbf{\$25,200}$; Return = $\frac{25,200 - 10,000}{10,000} = \mathbf{+152.0\%}$.

**Summary Table:**
| State | Prob | $S_T$ | Wealth (Shares) | Return (Shares) | Wealth (Calls+Cash) | Return (Calls+Cash) |
|:---|:---:|:---:|:---:|:---:|:---:|:---:|
| Weak | 0.20 | \$35 | \$7,000 | -30.0% | \$5,200 | -48.0% |
| Moderate | 0.40 | \$55 | \$11,000 | +10.0% | \$5,200 | -48.0% |
| Strong | 0.40 | \$75 | \$15,000 | +50.0% | \$25,200 | +152.0% |

**Part c. Expected return and expected terminal wealth:**
- **Strategy Shares:**
  $$E(W_{\text{shares}}) = 0.20(7,000) + 0.40(11,000) + 0.40(15,000) = 1,400 + 4,400 + 6,000 = \mathbf{\$11,800}$$
  $$E(r_{\text{shares}}) = 0.20(-30\%) + 0.40(10\%) + 0.40(50\%) = -6\% + 4\% + 20\% = \mathbf{18.00\%}$$
- **Strategy Calls and Cash:**
  $$E(W_{\text{calls}}) = 0.20(5,200) + 0.40(5,200) + 0.40(25,200) = 1,040 + 2,080 + 10,080 = \mathbf{\$13,200}$$
  $$E(r_{\text{calls}}) = 0.20(-48\%) + 0.40(-48\%) + 0.40(152\%) = -9.6\% - 19.2\% + 60.8\% = \mathbf{32.00\%}$$

**Part d. Assessment of Sigma's decision rule:**
- **Assessment:** Selecting the strategy based *solely* on higher expected return is **flawed and dangerous**.
- **Explanation:** 
  1. **Risk profile and leverage:** The Calls + Cash strategy provides significant embedded leverage on the upside (+152% in the strong state), which inflates its expected return (32% vs 18%).
  2. **Severe downside penalty:** In 60% of outcomes (both Weak and Moderate states), the calls expire worthless, and the investor suffers a severe loss of **-48%**. In the Moderate state, the shares deliver a positive return (+10%), while Calls + Cash delivers a heavy loss (-48%).
  3. **Risk aversion:** A risk-averse investor must evaluate the entire return distribution (risk, variance, and downside probability), not just the arithmetic mean. Under standard risk aversion, the severe left-tail outcomes of the options strategy may make it far less desirable than the equity strategy.

---
#### Financial Interpretation
Options create asymmetric leverage that dramatically alters higher return moments. Evaluating derivative strategies purely through expected return ignores downside crash risk and volatility.

### Question 9 (Easy)

Alpha has a bond with a face value of 2,000, a 4.5% annual coupon, and three years remaining to maturity. Its YTM is 6%.

a. List its remaining payments and calculate its price.

b. Explain whether it trades at a premium, at par, or at a discount.

---


### Solution to Question 9

#### Detailed Solution

**Given:**
- Face value $M = 2,000$
- Annual coupon rate $= 4.5\%$
- Annual coupon payment $C = 2,000 \times 4.5\% = 90$
- Maturity $T = 3$ years
- Yield to maturity $y = 6\% = 0.06$

**Part a. List remaining payments and calculate price:**
- **Cash flow schedule:**
  - Year 1: $C_1 = 90$
  - Year 2: $C_2 = 90$
  - Year 3: $C_3 + M = 90 + 2,000 = 2,090$
- **Bond price formula:**
  $$P = \frac{C}{1+y} + \frac{C}{(1+y)^2} + \frac{C+M}{(1+y)^3}$$
  $$P = \frac{90}{1.06} + \frac{90}{(1.06)^2} + \frac{2,090}{(1.06)^3}$$
  $$P = 84.905660 + 80.099680 + 1,754.804302 = \mathbf{1,919.809642}$$
  - **Price:** **USD 1,919.81**

**Part b. Premium, par, or discount:**
- **Classification:** The bond trades at a **discount**.
- **Explanation:** Whenever a bond's yield to maturity exceeds its coupon rate ($y = 6\% > \text{coupon rate} = 4.5\%$), the present value of its future payments is less than its par value ($P = 1,919.81 < 2,000$). Investors require a 6% return, so they will only purchase the bond if they receive a capital gain at maturity in addition to the 4.5% coupon stream.

---
#### Financial Interpretation
The relationship between coupon rate and YTM dictates bond pricing relative to par: Coupon Rate < YTM $\iff$ Price < Par (Discount); Coupon Rate = YTM $\iff$ Price = Par; Coupon Rate > YTM $\iff$ Price > Par (Premium).


### Question 10 (Easy)

Beta's bond has a face value of 1,000, an 8% coupon paid semiannually, and three years remaining. Its nominal annual YTM is 6%, compounded semiannually.

a. Calculate its price.

b. Gamma uses six payments of 80 and a discount rate of 6% per payment period, with 1,000 repaid at the end. Identify the errors in Gamma's calculation.

---


### Solution to Question 10

#### Detailed Solution

**Given:**
- Face value $M = 1,000$
- Coupon rate $= 8\%$ paid semiannually
- Semiannual coupon payment $C = 1,000 \times (8\% / 2) = 40$
- Maturity $= 3$ years $\implies n = 3 \times 2 = 6$ semiannual periods
- Quoted nominal annual YTM $= 6\%$, compounded semiannually
- Semiannual discount rate $r = 6\% / 2 = 3\% = 0.03$

**Part a. Calculate the bond's price:**
- **Pricing formula:**
  $$P = \sum_{t=1}^{6} \frac{C}{(1+r)^t} + \frac{M}{(1+r)^6} = C \left[\frac{1 - (1+r)^{-6}}{r}\right] + \frac{M}{(1+r)^6}$$
  $$P = 40 \left[\frac{1 - (1.03)^{-6}}{0.03}\right] + \frac{1,000}{(1.03)^6}$$
  $$P = 40 \times 5.4171914 + 1,000 \times 0.8374843 = 216.687658 + 837.484256 = \mathbf{1,054.171914}$$

### Solution to Question 22

#### Detailed Solution

**Given:**

- Chi: 3-year bond, face 1,000, annual coupon 40, price $P_{\text{Chi}} = 980$.

  $$980 = \frac{40}{1.03} + \frac{40}{(1.04)^2} + \frac{1,040}{(1 + s_3)^3}$$
  $$980 = 38.834951 + 36.982249 + \frac{1,040}{(1 + s_3)^3}$$
  $$980 - 75.817200 = \frac{1,040}{(1 + s_3)^3} \implies 904.182800 = \frac{1,040}{(1 + s_3)^3}$$
  $$(1 + s_3)^3 = \frac{1,040}{904.182800} = 1.150210$$

### Question 24 (Hard)

Beta buys a three-year bond with face value 1,000 and a 6% annual coupon when its YTM is 5%. One year later, Beta receives the coupon and immediately sells the bond. The sale price excludes the coupon just received. At that date the bond's YTM has the following distribution.

| State | Probability | YTM at sale |
|---|---:|---:|
| Lower yields | 0.25 | 3% |
| Unchanged yields | 0.50 | 5% |
| Higher yields | 0.25 | 8% |

a. Find the purchase price.

b. Find the sale price and one-year holding-period return in each state.

c. Calculate the expected holding-period return and its standard deviation.

d. Assess the claim: “Buying at a 5% YTM makes Beta's one-year return 5%, regardless of the sale-date yield.”

---


### Solution to Question 24

#### Detailed Solution

**Given:**
- Face value $= 1,000$, maturity $= 3$ years, coupon $= 6\% \implies C = 60$.
- Initial YTM $= 5\%$.
- Held for 1 year: receives Year 1 coupon (60), then sells bond with 2 years remaining.

**Part a. Purchase price ($P_0$):**
$$P_0 = \frac{60}{1.05} + \frac{60}{(1.05)^2} + \frac{1,060}{(1.05)^3} = 57.142857 + 54.421769 + 915.667854 = \mathbf{1,027.232480} \quad (\text{USD } 1,027.23)$$

**Part b. Sale prices and 1-year holding-period returns (HPR):**
At $t=1$, the bond has 2 years remaining. The sale price at yield $y$ is:
$$P_1(y) = \frac{60}{1+y} + \frac{1,060}{(1+y)^2}, \quad \text{HPR} = \frac{P_1 + 60 - P_0}{P_0}$$
- **State 1: Lower yields ($y = 3\%$, prob = 0.25):**
  $$P_1(0.03) = \frac{60}{1.03} + \frac{1,060}{(1.03)^2} = 58.252427 + 999.151664 = \mathbf{1,057.404091}$$
  $$\text{HPR}_1 = \frac{1,057.404091 + 60 - 1,027.232480}{1,027.232480} = \frac{90.171611}{1,027.232480} = \mathbf{+8.778111\%}$$
- **State 2: Unchanged yields ($y = 5\%$, prob = 0.50):**
  $$P_1(0.05) = \frac{60}{1.05} + \frac{1,060}{(1.05)^2} = 57.142857 + 961.451247 = \mathbf{1,018.594104}$$
  $$\text{HPR}_2 = \frac{1,018.594104 + 60 - 1,027.232480}{1,027.232480} = \frac{51.361624}{1,027.232480} = \mathbf{+5.000000\%}$$
- **State 3: Higher yields ($y = 8\%$, prob = 0.25):**
  $$P_1(0.08) = \frac{60}{1.08} + \frac{1,060}{(1.08)^2} = 55.555556 + 908.779149 = \mathbf{964.334705}$$
  $$\text{HPR}_3 = \frac{964.334705 + 60 - 1,027.232480}{1,027.232480} = \frac{-2.897775}{1,027.232480} = \mathbf{-0.282095\%}$$

**Part c. Expected HPR and standard deviation:**
- **Expected Return:**
  $$E(\text{HPR}) = 0.25(8.778111\%) + 0.50(5.000000\%) + 0.25(-0.282095\%)$$
  $$E(\text{HPR}) = 2.194528\% + 2.500000\% - 0.070524\% = \mathbf{4.624004\%} \quad (\approx 4.62\%)$$
- **Variance and Standard Deviation:**
  $$\sigma^2 = 0.25(8.778111 - 4.624004)^2 + 0.50(5.000000 - 4.624004)^2 + 0.25(-0.282095 - 4.624004)^2$$
  $$\sigma^2 = 0.25(4.154107)^2 + 0.50(0.375996)^2 + 0.25(-4.906099)^2 = 4.314144 + 0.070686 + 6.017452 = 10.402282 \; (\%^2)$$
  $$\sigma = \sqrt{10.402282} = \mathbf{3.225258\%} \quad (\approx 3.23\%)$$

**Part d. Assessment of the claim:**
- **Assessment:** The claim is **false**.
- **Explanation:** Buying a bond at a 5% YTM guarantees a 5% return *only* if the bond is held for exactly one period and yields do not change, or if held to maturity with coupons reinvested at 5%. If sold after one year, the investor faces **interest rate price risk**. If yields rise to 8%, capital losses depress the one-year return to -0.28%; if yields fall to 3%, capital gains boost the return to +8.78%. Only when the yield remains unchanged at 5% is the one-year return equal to the 5% YTM.

---
#### Financial Interpretation
Holding-period return over horizons shorter than maturity is uncertain and driven by capital gains/losses from interest rate fluctuations. YTM equals holding-period return only if the yield curve is unchanged.

### Question 25 (Easy)

Alpha's bond costs 1,000. Its remaining payments have the following present values.

### Solution to Question 25

#### Detailed Solution

**Given:**
- Price $P = 1,000$
- $PV_1 = 40 \implies w_1 = 40 / 1,000 = 0.040$
- $PV_2 = 38 \implies w_2 = 38 / 1,000 = 0.038$
- $PV_3 = 922$

**Part a. Missing weight and total weight:**
- **Missing weight ($w_3$):**

**Part b. Macaulay duration and explanation:**
- **Macaulay Duration Formula:**
  $$D_{\text{mac}} = \sum_{t=1}^{3} t \cdot w_t = 1(0.040) + 2(0.038) + 3(0.922)$$

Only for a zero-coupon bond does Macaulay duration equal nominal maturity.

---

Macaulay duration measures the economic life of a bond. Intermediate coupon payments reduce duration relative to maturity, shortening price sensitivity to interest rates.

### Solution to Question 26

#### Detailed Solution

**Given:**
- $D_{\text{mac}} = 4.2$ years for both Beta and Gamma.
- Beta: Annual coupons, quoted annual YTM $= 8\%$ effective ($m = 1$).
- Gamma: Semiannual coupons, quoted nominal annual YTM $= 8\%$, compounded semiannually ($m = 2, y/m = 4\%$).

- **Modified Duration Formula:**
  $$D^* = \frac{D_{\text{mac}}}{1 + y/m}$$
- **Beta ($m = 1$):**
  $$D^*_{\text{Beta}} = \frac{4.2}{1 + 0.08} = \frac{4.2}{1.08} = \mathbf{3.888889 \text{ years}}$$

---
#### Financial Interpretation
Modified duration directly measures price sensitivity per unit change in quoted yield. Semiannual compounding reduces the period discount rate, resulting in higher modified duration for a given Macaulay duration.

| Annual convexity | 54.476335 years² |

Its repriced values are 966.336276 at a 4% yield and 791.004552 at a 7% yield.

a. Estimate the price at each new yield using duration alone and then duration plus convexity.

### Solution to Question 32

#### Detailed Solution

**Given:**
- $y_0 = 5.5\% = 0.055$, $P_0 = 873.308680$, $D^* = 6.673273$, $C = 54.476335$.
- Shock 1: $y = 4.0\% \implies \Delta y = -0.0150$. Actual price $= 966.336276$.
- Shock 2: $y = 7.0\% \implies \Delta y = +0.0150$. Actual price $= 791.004552$.

**Part a. Price estimates:**
- **Formulas:**
  - Duration only: $P_{\text{dur}} = P_0 (1 - D^* \Delta y)$
  - Duration + Convexity: $P_{\text{both}} = P_0 [1 - D^* \Delta y + \frac{1}{2} C (\Delta y)^2]$
- **At $y = 4.0\%$ ($\Delta y = -0.015$):**
  - $\Delta P_{\text{dur}} / P_0 = -6.673273 \times (-0.015) = +0.10009910 \implies P_{\text{dur}} = 873.308680 \times 1.10009910 = \mathbf{960.726052}$
  - Convexity term: $\frac{1}{2} (54.476335)(-0.015)^2 = +0.00612859$
  - $P_{\text{both}} = 873.308680 \times (1 + 0.10009910 + 0.00612859) = \mathbf{966.078235}$
- **At $y = 7.0\%$ ($\Delta y = +0.015$):**
  - $\Delta P_{\text{dur}} / P_0 = -6.673273 \times (+0.015) = -0.10009910 \implies P_{\text{dur}} = 873.308680 \times 0.89990090 = \mathbf{785.891308}$

- **At $y = 4.0\%$ (Actual $= 966.336276$):**
  - Duration only: $960.726052 - 966.336276 = \mathbf{-5.610224} \; (-\mathbf{0.6424\%}\text{ of } P_0)$

The positive convexity of option-free bonds means that price increases by more when yields fall than it decreases when yields rise. Duration-only estimates consistently underestimate actual prices for both upward and downward yield shifts.

### Question 35 (Average)

Six months ago, Pi's fund matched the present value and duration of its coupon-bond assets to a single fixed liability. Since then, short-maturity yields have risen by 50 basis points and long-maturity yields have fallen by 25 basis points. The fund has not traded. Its manager proposes no review because the original durations matched.

a. Assess this proposal. What needs to be checked now?

b. Pi could instead afford a default-free, noncallable zero that pays exactly the liability amount on its due date, in the same currency. Would that replacement require the same rebalancing for rate changes? Explain.

---


### Solution to Question 35

#### Detailed Solution

**Part a. Assessment of proposal and necessary review:**
- **Assessment:** The proposal of "no review" is **seriously flawed and unacceptable**.
- **What needs to be checked immediately:**
  1. **Non-parallel yield shift:** Short yields rose by 50 bps while long yields fell by 25 bps (a yield curve flattening / twist). Classical duration matching protects only against *parallel* yield shifts. A non-parallel shift alters asset values and liability present value differently, creating an asset-liability mismatch.
  2. **Passage of time:** Over the past 6 months, the liability's duration decreased by exactly 0.5 years. Coupon bonds with different maturities and cash flow schedules do not experience duration declines at the same rate.
  3. **Convexity / duration drift:** Changes in yield levels alter the modified durations of the bonds themselves.
  - **Action required:** Re-evaluate the present value and duration of both assets and liabilities to rebalance the portfolio back into immunization.

**Part b. Replacement with a matching zero-coupon bond:**
- **Assessment:** A dedicated matching zero-coupon bond **would NOT require rebalancing** for interest rate changes.
- **Explanation:** If the fund holds a default-free, noncallable zero maturing exactly on the liability due date with face value equal to the liability, the fund achieves **cash-flow matching (dedication)**. The cash inflow on the due date matches the cash outflow dollar-for-dollar, regardless of what interest rates do in the interim. There is zero price risk at maturity and zero reinvestment risk.

---
#### Financial Interpretation
Duration matching requires continuous dynamic rebalancing due to non-parallel shifts and duration drift. Cash flow matching with zero-coupon bonds provides passive, perfect immunization.

### Question 53 (Hard)

Eta forms an equally weighted portfolio of $n$ stocks. Every stock has annual SD of 30%, and the covariance between every distinct pair is 0.0100 in decimal-return units. These inputs remain unchanged as more stocks are added.

a. Find portfolio variance and SD for $n=10$.

b. Find the limiting SD as $n$ grows without bound.

c. What is the smallest whole number of stocks needed to obtain SD no greater than 12%?

d. Could adding stocks achieve an SD of 9% under these assumptions? Explain.

---


### Solution to Question 53

#### Detailed Solution

**Given:** Equally weighted portfolio of $n$ stocks. $\sigma_i = 30\% \implies \sigma_i^2 = 0.0900$. Distinct pair covariance $\text{Cov} = 0.0100$.

**Part a. Portfolio variance and SD for $n = 10$:**
$$\sigma_p^2 = \frac{1}{n} \bar{\sigma}^2 + \frac{n-1}{n} \overline{\text{Cov}} = \frac{1}{10}(0.0900) + \frac{9}{10}(0.0100) = 0.0090 + 0.0090 = \mathbf{0.018000}$$
$$\sigma_p = \sqrt{0.018000} = \mathbf{13.416408\%} \quad (\approx 13.42\%)$$

**Part b. Limiting SD as $n \to \infty$:**
$$\lim_{n \to \infty} \sigma_p^2 = \overline{\text{Cov}} = 0.0100 \implies \lim_{n \to \infty} \sigma_p = \sqrt{0.0100} = 0.1000 = \mathbf{10.0000\%}$$

**Part c. Minimum $n$ for $\sigma_p \le 12\%$:**
$$\sigma_p^2 \le (0.12)^2 = 0.0144$$
$$\frac{0.0900 - 0.0100}{n} + 0.0100 \le 0.0144 \implies \frac{0.0800}{n} \le 0.0044 \implies n \ge \frac{0.0800}{0.0044} = \frac{800}{44} \approx 18.18$$
- Smallest whole number of stocks: **n = 19**.
  *(Check: at $n=18$, $\sigma_p = 12.02\%$; at $n=19$, $\sigma_p = 11.92\% \le 12\%$.)*

**Part d. Can adding stocks achieve an SD of 9%?**
- **No.** The undiversifiable systematic risk establishes a strict mathematical floor at $\lim_{n \to \infty} \sigma_p = 10.0\%$. Because all pair covariances are positive ($0.0100$), diversification eliminates only unique/firm-specific variance ($0.08/n$), leaving the common systematic covariance untouched. A standard deviation of 9% is impossible under these assumptions.

---
#### Financial Interpretation
Diversification eliminates firm-specific risk as $n \to \infty$, but cannot eliminate common systematic covariance.

### Solution to Question 61

#### Detailed Solution

**Given:**
- $\sigma_L^2 = 0.040000, \sigma_M^2 = 0.010000, \text{Cov}_{LM} = 0.003000$.
- $v(w) = 0.04 w^2 + 0.01 (1 - w)^2 + 0.006 w (1 - w)$.

**Part a. Complete the derivative and find global minimum-variance weights:**
- **Derivative expansion:**
  $$\frac{dv}{dw} = 2(0.04)w - 2(0.01)(1 - w) + 0.006(1 - 2w)$$
  $$\frac{dv}{dw} = 0.08w - 0.02 + 0.02w + 0.006 - 0.012w = \mathbf{0.088w - 0.014}$$
- **First-order condition ($dv/dw = 0$):**
  $$0.088w - 0.014 = 0 \implies w^* = \frac{0.014}{0.088} = \frac{14}{88} = \mathbf{\frac{7}{44} \approx 0.159091}$$

### Question 68 (Average)

Delta and Epsilon use mean–variance utility, with $A=2$ and $A=6$. They share forecasts and access to the same risky assets. Initially, both may short sell and lend or borrow at the same risk-free rate. Assume a unique tangency portfolio with a positive risk premium.

### Solution to Question 68

#### Detailed Solution

**Part a. Common vs. differing portfolio characteristics:**
- **What is common:** Under the **Separation Property (Two-Fund Separation)**, both Delta and Epsilon hold the **exact same risky portfolio** (the unique tangency portfolio $P$). Portfolio $P$ is determined purely by asset returns, volatilities, and covariances, independent of investor risk preferences.
- **What differs:** Their **capital allocation** $y^*$ between portfolio $P$ and the risk-free asset differs based on risk aversion:
  $$y^* = \frac{E(r_P) - r_f}{A \sigma_P^2}$$
  Delta ($A=2$) chooses three times the risky exposure of Epsilon ($A=6$): $y^*_{\Delta} = 3 y^*_E$.

**Part b. Binding borrowing constraint on Delta:**
- **Assessment:** **Yes**, Delta's new optimal portfolio must be **100% in the original tangency portfolio $P$ ($y = 1.0$)**.

Delta does *not* revise the internal risky asset mix because no other risky combination achieves a higher Sharpe ratio.

Because Delta still has unrestricted access to short selling, Delta continues to hold the superior unconstrained portfolio, so their risky portfolios diverge.

---
#### Financial Interpretation
Two-fund separation holds only when all investors face identical access to assets and borrowing. Asymmetric portfolio constraints break the separation property.

### Question 83 (Average)

A historical regression estimates Phi's monthly excess return as

$$\hat{R}_{\mathrm{Phi},t}=0.003+1.2R_{M,t}.$$

Phi has nonzero residual variance. Assess each claim:

a. “The positive fitted intercept disproves CAPM and establishes a positive future alpha.”

b. “If Phi is correctly priced, it belongs on the capital market line (CML) at the horizontal coordinate 1.2.”

c. “A positive forecast alpha would guarantee a risk-free profit from buying Phi and hedging its market beta.”

---


### Solution to Question 83

#### Detailed Solution

**Given:** Single-index regression $\hat{R}_{\text{Phi},t} = 0.003 + 1.2 R_{M,t}$ with $\sigma_e > 0$.

**Part a. "The positive fitted intercept disproves CAPM and establishes a positive future alpha."**
- **Assessment:** **False**.
- **Explanation:** A positive sample intercept in historical data is an empirical estimate subject to sampling error; it does not disprove CAPM (it may be statistically indistinguishable from zero or driven by a lucky sample period). Furthermore, past sample performance does not establish a positive *future* expected alpha; in competitive equilibrium, mispricings are eliminated.

**Part b. "If Phi is correctly priced, it belongs on the capital market line (CML) at the horizontal coordinate 1.2."**
- **Assessment:** **False**.
- **Correction:** The Capital Market Line (CML) plots expected return against **total risk ($\sigma$)**, NOT beta! Only the market portfolio and combinations of the market portfolio and the risk-free asset lie on the CML. Individual stocks have non-zero idiosyncratic risk ($\sigma_e > 0$), so they lie strictly **below the CML**. The line that plots expected returns against beta (horizontal coordinate 1.2) is the **Security Market Line (SML)**, not the CML.

**Part c. "A positive forecast alpha would guarantee a risk-free profit from buying Phi and hedging its market beta."**
- **Assessment:** **False**.
- **Explanation:** Buying Phi and shorting $1.2$ units of the market eliminates systematic market risk, but leaves **residual idiosyncratic risk** ($\sigma_e > 0$) unhedged:
  $$r_{\text{hedged}} = \alpha_{\text{Phi}} + e_{\text{Phi}}$$
  The profit is uncertain and subject to firm-specific shocks. It is an active speculative arbitrage bet, NOT a risk-free profit.

---
#### Financial Interpretation
Distinguishing the CML (total risk, efficient portfolios) from the SML (systematic beta, all securities) is a cornerstone of modern portfolio theory.

### Question 87 (Easy)

Mu and Nu are correctly priced by a two-factor model. The risk-free rate is 3%.

| Factor | Annual risk premium | Mu loading | Nu loading |
|---|---:|---:|---:|

a. Calculate each asset's expected total return.

b. Calculate the expected-return difference, Nu minus Mu, and identify the factor contribution that accounts for it.

---


### Solution to Question 87

#### Detailed Solution

**Given:** Two-factor model: $r_f = 3\% = 0.03$.
- GDP growth factor premium $\lambda_{\text{GDP}} = 5\% = 0.05$
- Inflation factor premium $\lambda_{\text{inf}} = -2\% = -0.02$
- Mu: $\beta_{\text{GDP}} = 1.1, \beta_{\text{inf}} = 0.4$
- Nu: $\beta_{\text{GDP}} = 1.1, \beta_{\text{inf}} = -0.6$

$$E(r_i) = r_f + \beta_{i,\text{GDP}} \lambda_{\text{GDP}} + \beta_{i,\text{inf}} \lambda_{\text{inf}}$$

- **Attribution:**
  Both assets have identical exposure to GDP growth ($\beta_{\text{GDP}} = 1.1$). The entire +2.00% difference is accounted for by the **inflation factor**:
  $$[\beta_{\text{Nu},\text{inf}} - \beta_{\text{Mu},\text{inf}}] \times \lambda_{\text{inf}} = [-0.6 - 0.4] \times (-2\%) = -1.0 \times (-2\%) = \mathbf{+2.0000\%}$$
  Because inflation carries a negative risk premium ($-2\%$, meaning inflation is an unfavorable state of the world where hedging is prized), an asset that has a negative loading on inflation (Nu) performs worse in inflation, providing no hedge, and thus requires a higher expected return (+1.2% premium vs -0.8% penalty for Mu).

### Question 90 (Easy)

Assess the following claims.

a. Tau: “APT needs the same efficient market portfolio and mean–variance investor assumptions as CAPM.”

b. Upsilon: “Adding a second factor increases a return regression's $R^2$, so that factor must earn a risk premium.” Explain also what APT leaves unspecified about the factors.

### Solution to Question 90

#### Detailed Solution

**Part a. Assessment of Tau's claim:**
- *"APT needs the same efficient market portfolio and mean–variance investor assumptions as CAPM."*
- **Assessment:** **False**.
- **Explanation:** 
  1. **No Market Portfolio:** APT does *not* require identifying or holding the market portfolio, avoiding Roll's critique.
  2. **No Utility Specification:** APT does not require mean-variance preferences or normally distributed asset returns. It relies only on the **no-arbitrage condition** across well-diversified portfolios and a linear factor structure.

**Part b. Assessment of Upsilon's claim and what APT leaves unspecified:**
- *"Adding a second factor increases a return regression's $R^2$, so that factor must earn a risk premium."*
- **Assessment:** **False**.

  1. The **number** of factors $K$.

APT is more theoretically robust than CAPM by avoiding assumptions about the unobservable market portfolio, but leaves empirical factor identification open.

- Zeta: $\beta_1 = 0.4, \beta_2 = 1.2, \sigma_{eZ}^2 = 0.014400$
- Uncorrelated residuals.
- Target: Portfolio Factor 1 loading $= 1.0$, Factor 2 loading $= 0.0$.

**Part a. Find weights and monetary positions:**
- System of equations for $w_{\Omega}$ and $w_Z$:
  $$1.1 w_{\Omega} + 0.4 w_Z = 1.0$$
  $$0.6 w_{\Omega} + 1.2 w_Z = 0.0 \implies 1.2 w_Z = -0.6 w_{\Omega} \implies w_Z = -0.5 w_{\Omega}$$
- Substitute into the first equation:
  $$1.1 w_{\Omega} + 0.4(-0.5 w_{\Omega}) = 1.0 \implies 1.1 w_{\Omega} - 0.2 w_{\Omega} = 1.0 \implies 0.9 w_{\Omega} = 1.0$$
  $$\mathbf{w_{\Omega} = \frac{10}{9} \approx 1.111111}, \quad \mathbf{w_Z = -\frac{5}{9} \approx -0.555556}$$
  $$\mathbf{w_{\text{cash}} = 1 - \frac{10}{9} - \left(-\frac{5}{9}\right) = 1 - \frac{5}{9} = \frac{4}{9} \approx 0.444444}$$
- **Monetary Positions (Wealth = 90,000):**
  - **Omega:** $90,000 \times (10/9) = \mathbf{+\text{USD } 100,000.00}$ (Long)
  - **Zeta:** $90,000 \times (-5/9) = \mathbf{-\text{USD } 50,000.00}$ (Short)
  - **Cash:** $90,000 \times (4/9) = \mathbf{+\text{USD } 40,000.00}$ (Lending)
  - *(Check sum: $100,000 - 50,000 + 40,000 = \$90,000.00$.)*

**Part b. Residual variance, residual SD, and total SD:**
- **Residual Variance:**
  $$\sigma_e^2(p) = w_{\Omega}^2 \sigma_{e\Omega}^2 + w_Z^2 \sigma_{eZ}^2 = \left(\frac{10}{9}\right)^2(0.008100) + \left(-\frac{5}{9}\right)^2(0.014400)$$
  $$\sigma_e^2(p) = \frac{100}{81}(0.008100) + \frac{25}{81}(0.014400) = 0.010000 + 0.00444444 = \mathbf{0.01444444}$$
- **Residual SD:**
  $$\sigma_e(p) = \sqrt{0.01444444} = \mathbf{12.018504\%} \quad (\approx 12.02\%)$$
- **Total Variance and Total SD:**
  $$\sigma_p^2 = \beta_1^2 \sigma^2(F_1) + \sigma_e^2(p) = (1.0)^2(0.16)^2 + 0.01444444 = 0.025600 + 0.01444444 = \mathbf{0.04004444}$$
  $$\sigma_p = \sqrt{0.04004444} = \mathbf{20.011108\%} \quad (\approx 20.01\%)$$

**Part c. Tracking error and risk-free status:**
- **Does it track Factor 1 without error?**
  **No.** The portfolio contains non-zero idiosyncratic risk (residual SD $= 12.02\%$). Thus, its realized return will deviate from Factor 1 due to firm-specific shocks.
- **Is it risk-free?**
  **No.** The portfolio has a total volatility of 20.01% (composed of both Factor 1 systematic risk and residual risk). It was constructed to have zero exposure to Factor 2, not zero risk overall.

---
#### Financial Interpretation
Hedging one systematic risk factor eliminates sensitivity to that specific macro variable, but leverage can magnify idiosyncratic volatility.


### Question 100 (Hard)

Rho and Sigma are evaluated using the same 60 monthly observations. Fund returns are after fees. The sample arithmetic means are:

| Series | Monthly mean |
|---|---:|
| Market excess return | 0.60% |
| SMB | 0.20% |
| HML | 0.30% |
| Risk-free return | 0.20% |
| Rho total return | 1.20% |
| Sigma total return | 1.08% |

Separate OLS regressions, each including an intercept, report:

| Fund | CAPM market beta | FF3 market loading | FF3 SMB loading | FF3 HML loading |
|---|---:|---:|---:|---:|
| Rho | 1.00 | 0.90 | 0.80 | 0.60 |
| Sigma | 0.95 | 1.00 | −0.20 | 0.10 |

a. Calculate each fund's zero-alpha benchmark total return and monthly alpha under CAPM.

b. Repeat under FF3.

c. Reconcile the change in each fund's alpha using the changes in market, size, and value contributions.

d. Rank the funds by estimated alpha under each model. Explain what the comparison shows about performance assessment.

---


### Solution to Question 100

#### Detailed Solution

**Given:** Sample means: Market excess $= 0.60\%$, SMB $= 0.20\%$, HML $= 0.30\%$, $r_f = 0.20\%$.
- Rho total return $= 1.20\% \implies$ Excess return $= 1.20\% - 0.20\% = 1.00\%$.
- Sigma total return $= 1.08\% \implies$ Excess return $= 1.08\% - 0.20\% = 0.88\%$.

**Part a. CAPM benchmark and alpha:**
- **Rho ($\beta_M = 1.00$):**
  $$r_{\text{bench}} = 0.20\% + 1.00(0.60\%) = \mathbf{0.8000\%}$$
  $$\alpha_{\text{CAPM}} = 1.20\% - 0.80\% = \mathbf{+0.4000\%}$$
- **Sigma ($\beta_M = 0.95$):**
  $$r_{\text{bench}} = 0.20\% + 0.95(0.60\%) = 0.20\% + 0.57\% = \mathbf{0.7700\%}$$
  $$\alpha_{\text{CAPM}} = 1.08\% - 0.77\% = \mathbf{+0.3100\%}$$

**Part b. Fama-French 3-Factor (FF3) benchmark and alpha:**
- **Rho ($\beta_M = 0.90, \beta_{\text{SMB}} = 0.80, \beta_{\text{HML}} = 0.60$):**
  $$R_{\text{bench}} = 0.90(0.60\%) + 0.80(0.20\%) + 0.60(0.30\%) = 0.54\% + 0.16\% + 0.18\% = 0.8800\%$$
  $$r_{\text{bench}} = 0.20\% + 0.88\% = \mathbf{1.0800\%}$$
  $$\alpha_{\text{FF3}} = 1.20\% - 1.08\% = \mathbf{+0.1200\%}$$
- **Sigma ($\beta_M = 1.00, \beta_{\text{SMB}} = -0.20, \beta_{\text{HML}} = 0.10$):**
  $$R_{\text{bench}} = 1.00(0.60\%) - 0.20(0.20\%) + 0.10(0.30\%) = 0.60\% - 0.04\% + 0.03\% = 0.5900\%$$
  $$r_{\text{bench}} = 0.20\% + 0.59\% = \mathbf{0.7900\%}$$
  $$\alpha_{\text{FF3}} = 1.08\% - 0.79\% = \mathbf{+0.2900\%}$$

**Part c. Reconcile alpha changes:**
- **Rho:** Alpha drops from $+0.40\%$ to $+0.12\%$ (a drop of $-0.28\%$).
  - Market effect: $(0.90 - 1.00) \times 0.60\% = -0.06\%$
  - Size effect: $0.80 \times 0.20\% = +0.16\%$
  - Value effect: $0.60 \times 0.30\% = +0.18\%$
  - Net benchmark increase: $-0.06\% + 0.16\% + 0.18\% = +0.28\%$, explaining the exact $0.28\%$ drop in alpha.
- **Sigma:** Alpha drops only slightly from $+0.31\%$ to $+0.29\%$ (a drop of $-0.02\%$).
  - Market effect: $(1.00 - 0.95) \times 0.60\% = +0.03\%$
  - Size effect: $-0.20 \times 0.20\% = -0.04\%$
  - Value effect: $0.10 \times 0.30\% = +0.03\%$
  - Net benchmark increase: $+0.03\% - 0.04\% + 0.03\% = +0.02\%$, explaining the $0.02\%$ drop.

**Part d. Ranking comparison and performance assessment:**
- **CAPM Ranking:** **Rho (+0.40%) > Sigma (+0.31%)**.
- **FF3 Ranking:** **Sigma (+0.29%) > Rho (+0.12%)**.
- **Performance Assessment Lesson:**
  The ranking **completely reverses** between the models! Under CAPM, Rho appears to have superior managerial skill because CAPM mistakes style returns for manager alpha. In reality, Rho was merely taking heavy systematic bets on small-cap stocks ($\beta_{\text{SMB}} = 0.80$) and value stocks ($\beta_{\text{HML}} = 0.60$). Once these passive factor risks are properly accounted for in the FF3 model, Rho's genuine alpha collapses to 0.12%, while Sigma generates more than double the true manager skill (+0.29%).

---
#### Financial Interpretation
Multi-factor performance attribution prevents mistaking rewarded risk factors for managerial alpha.
