## Lecture 3: Duration, Convexity and Immunization

All amounts are in USD. Unless stated otherwise, bonds are default-free and noncallable, coupons are annual, yields are annual effective rates, and valuation occurs immediately after a coupon payment. Ignore taxes and transaction costs. Yield shocks are instantaneous: no time passes and no payment occurs during a shock.

Duration is reported in years and annual convexity in years squared. Use decimal yield changes in calculations; one basis point is 0.0001. Retain unrounded intermediate values and report durations and convexities to four decimal places. Where rounded inputs are supplied, use their displayed precision.
