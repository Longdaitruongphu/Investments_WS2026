### Bond Prices and Yields
*   The nominal risk-free interest rate is approximately the sum of
    *   A real risk-free rate of return.
    *   Compensation for expected inflation.
*   A bond's required yield also reflects its default risk, liquidity, tax treatment, and call risk.

$$ P_B = \sum_{t=1}^{T} \frac{C}{(1+r)^t} + \frac{Par Value}{(1+r)^T} $$

$$ P_B = C * \frac{1}{r} \left[1 - \frac{1}{(1+r)^T}\right] + \frac{Par Value}{(1+r)^T} $$

*   Theta issues a 10-year bond with a par value of \$1,000 and a 6% annual coupon rate, paid semiannually. If its yield to maturity is 8% per year, compounded semiannually, what is the bond's price?

$$ P_B = \frac{1,000 \cdot 6\%}{2} \cdot \frac{1}{8\%/2} \left[1 - \frac{1}{(1+8\%/2)^{20}}\right] + \frac{1,000}{(1+8\%/2)^{20}} $$

$$P_B = \$864.10$$

```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display

# --- 1. Core Calculation Function ---
def calculate_bond_price(face_value, coupon_rate, market_rate, years_to_maturity, freq=1):
    """
    Calculates the price of a bond.
    
    Args:
        face_value (float): The par value of the bond.
        coupon_rate (float): The annual coupon rate (in decimal form, e.g., 0.08 for 8%).
        market_rate (float): Nominal annual YTM, compounded at the coupon frequency (in decimal form).
        years_to_maturity (int): The number of years until the bond matures.
        freq (int): Number of coupon payments per year.
    """
    # Adjust rates and periods for payment frequency
    periods = years_to_maturity * freq
    coupon_payment = (face_value * coupon_rate) / freq
    r = market_rate / freq

    # Handle edge case for zero market rate
    if r == 0:
        pv_coupons = coupon_payment * periods
    else:
        pv_coupons = coupon_payment * ((1 - (1 + r)**-periods) / r)
    
    pv_face_value = face_value / (1 + r)**periods
    
    return pv_coupons + pv_face_value

# --- 2. Plotting Function ---
def generate_bond_plot(coupon_rate, years_to_maturity, current_market_rate, coupon_frequency=1):
    """
    Generates and displays the bond price vs. interest rate plot.
    """
    # Convert slider percentages to decimals for calculation
    coupon_r = coupon_rate / 100
    market_r = current_market_rate / 100
    face_value = 1000

    # --- Data for the curve ---
    # Create a range of interest rates for the x-axis
    rates_range = np.linspace(0.001, 0.20, 200) # From 0.1% to 20%
    # Calculate bond price for each interest rate in the range
    prices = [calculate_bond_price(face_value, coupon_r, r, years_to_maturity, freq=coupon_frequency) for r in rates_range]
    
    # --- Calculate the price for the single interactive point ---
    current_price = calculate_bond_price(face_value, coupon_r, market_r, years_to_maturity, freq=coupon_frequency)

    # --- Plotting ---
    fig, ax = plt.subplots(figsize=(8, 5))
    
    # Plot the main curve
    ax.plot(rates_range * 100, prices, color='deepskyblue', linewidth=2.5)

    # Plot the interactive point
    ax.plot(current_market_rate, current_price, 'o', color='red', markersize=8, zorder=5)

    # Plot the dashed lines to the axes
    ax.plot([current_market_rate, current_market_rate], [0, current_price], 'r--', alpha=0.7)
    ax.plot([0, current_market_rate], [current_price, current_price], 'r--', alpha=0.7)

    # Add text annotation for the point
    ax.text(current_market_rate + 0.5, current_price + 50, 
            f'({current_market_rate:.1f}%, ${current_price:,.2f})', 
            fontsize=12, color='black')

    # --- Formatting ---
    frequency_label = {1: 'annual', 2: 'semiannual', 4: 'quarterly'}[coupon_frequency]
    ax.set_title(f'Bond Price vs. Market Rate\n{coupon_rate:g}% coupon, {years_to_maturity} years, {frequency_label} payments', fontsize=16)
    ax.set_xlabel('Nominal Annual Interest Rate (%)', fontsize=12)
    ax.set_ylabel('Bond Price ($)', fontsize=12)
    ax.set_xlim(0, 20)
    ax.set_ylim(bottom=0, top=max(prices) * 1.1) # Dynamic top limit
    ax.grid(True, linestyle='--', alpha=0.6)
    
    plt.show()

# --- 3. Interactive Widgets Setup ---

# Define the widgets for user input
coupon_rate_widget = widgets.FloatSlider(
    value=5.0, min=0, max=20, step=0.5, 
    description='Coupon Rate (%):', style={'description_width': 'initial'}, continuous_update=False
)
years_widget = widgets.IntSlider(
    value=30, min=1, max=50, step=1, 
    description='Years to Maturity:', style={'description_width': 'initial'}, continuous_update=False
)
market_rate_widget = widgets.FloatSlider(
    value=5.0, min=0.1, max=20, step=0.1, 
    description='Market Interest Rate (%):', style={'description_width': 'initial'}, continuous_update=True,
)

bond_price_frequency_widget = widgets.Dropdown(
    options=[('Annual', 1), ('Semiannual', 2), ('Quarterly', 4)],
    value=1,
    description='Coupon Frequency:', style={'description_width': 'initial'}
)

# Capture each update in this demonstration's own output widget.
bond_price_output = widgets.interactive_output(
    generate_bond_plot,
    {
        'coupon_rate': coupon_rate_widget,
        'years_to_maturity': years_widget,
        'current_market_rate': market_rate_widget,
        'coupon_frequency': bond_price_frequency_widget
    }
)

# Arrange the widgets in a box
bond_price_controls = widgets.VBox([
    widgets.HTML("<h3>Set Bond Characteristics:</h3>"),
    coupon_rate_widget,
    bond_price_frequency_widget,
    years_widget,
    market_rate_widget
])

# Display the controls and the output area
display(bond_price_controls, bond_price_output)
```

### Accrued Interest and Quoted Bond Prices
*   If a bond is purchased between coupon payments, the buyer must pay the seller for accrued interest.
    *   It is the prorated share of the upcoming semiannual coupon.
    *   The quoted price does not include the interest that accrues between coupon payment dates.

$$ \text{Accrued interest} = \frac{\text{Annual coupon payment}}{2} \times \frac{\text{Days since last coupon payment}}{\text{Days separating coupon payments}} $$

*   Gamma buys a bond 45 days after its last coupon payment. The bond has a \$1,000 par value, a 6% annual coupon paid semiannually, and 180 days between coupon payments. How much accrued interest does Gamma pay the seller?

$$ \text{Accrued Interest} = \text{Coupon payment} \times \frac{\text{Days since last coupon payment}}{\text{Days separating coupon payments}} $$

$$ \text{Accrued Interest} = \frac{1,000 \cdot 6\%}{2} \cdot \frac{45}{180} = \$7.50 $$

*   Stocks trade without a separate "accrued dividend" payment. Investors who buy before the ex-dividend date and hold until that date receive the upcoming dividend; buyers on or after that date do not. All else equal, the share price falls by the dividend amount on the ex-dividend date.

### Yield to Maturity
*   The interest rate that makes the present value of a bond's payments equal to its price.

*   A measure of the average rate of return that will be earned on a bond if it is bought now and held until maturity.
    
    *   Assumption: All coupons can be reinvested at YTM
    
### Current yield
*   The bond's annual coupon payment divided by its price.

*   Lambda buys a 5-year bond for $\$1,085.30$. Its par value is $\$1,000$, and it pays $\$40$ every six months. Find its current yield, nominal annual yield to maturity, and effective annual yield.

$$ \text{Current Yield} = \frac{\text{Annual Coupon}}{\text{Price}} $$

$$ \text{Current Yield} = \frac{40 \cdot 2}{1,085.30} = 7.37\% $$

$$ 1,085.30 \approx 40 \cdot \frac{1}{r/2} \left[1 - \frac{1}{(1+r/2)^{10}}\right] + \frac{1,000}{(1+r/2)^{10}} $$

$$ r/2 \approx 3\% \quad \Rightarrow \quad r \approx 6\% \quad \Rightarrow \quad \text{Effective annual yield} = (1.03)^2 - 1 = 6.09\% $$

```python
#!pip install ipywidgets matplotlib

import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, clear_output

# ----------------------------
# COUPON-DATE BOND VALUES
# ----------------------------
def bond_price(remaining_maturity, coupon_rate, face_value, ytm, coupon_freq):
    """
    Price on a coupon date, excluding the coupon due on that date.
    At maturity, returns the principal to be repaid.
    """
    # Maturities are sampled at coupon dates; round away floating-point noise.
    periods = int(round(remaining_maturity * coupon_freq))
    if periods == 0:
        return face_value  # At maturity
    
    period_yield = ytm / coupon_freq
    coupon_pmt = (face_value * coupon_rate) / coupon_freq
    
    # Discount each remaining payment; this also works when YTM is zero.
    discount_factors = (1 + period_yield) ** (-np.arange(1, periods + 1, dtype=float))
    pv_coupons = coupon_pmt * discount_factors.sum()
    pv_face = face_value * discount_factors[-1]
    
    return pv_coupons + pv_face

def plot_bond_price_dynamics(face_value, ytm_percent, coupon1_percent, coupon2_percent, coupon_freq):
    """
    Plots exact coupon-date prices as maturity approaches, at a constant YTM.
    """
    # Convert percentages to decimals
    ytm = ytm_percent / 100
    coupon1 = coupon1_percent / 100
    coupon2 = coupon2_percent / 100
    
    # Include every coupon date for the selected payment frequency.
    maturities = np.arange(5 * coupon_freq, -1, -1) / coupon_freq
    
    # Compute discrete prices for each coupon rate
    bond_prices_1 = [bond_price(t, coupon1, face_value, ytm, coupon_freq) for t in maturities]
    bond_prices_2 = [bond_price(t, coupon2, face_value, ytm, coupon_freq) for t in maturities]
    
    # Connect the calculated values without interpolating extra prices.
    plt.figure(figsize=(8, 5), dpi=100)
    plt.plot(maturities, bond_prices_1, marker='o', markersize=4,
             label=f'{coupon1_percent:g}% Coupon Bond (YTM {ytm_percent:g}%)')
    plt.plot(maturities, bond_prices_2, marker='o', markersize=4,
             label=f'{coupon2_percent:g}% Coupon Bond (YTM {ytm_percent:g}%)')

    # Flip axis: 5 years on left to 0 years on right
    plt.gca().invert_xaxis()
    plt.title("Bond Prices on Coupon Dates")
    plt.xlabel("Remaining Maturity (Years)")
    plt.ylabel("Bond Price ($)")
    plt.grid(True, alpha=0.3, linestyle=':')
    plt.legend()
    plt.figtext(0.5, 0.02, "Prices exclude the coupon due; the maturity value is principal. Lines connect the points.",
                ha='center', fontsize=9)
    plt.tight_layout(rect=(0, 0.05, 1, 1))
    plt.show()

# -------------------------
# WIDGETS FOR INTERACTIVITY
# -------------------------
face_value_widget = widgets.FloatText(
    value=1000,
    description='Face Value:'
)
ytm_widget = widgets.FloatText(
    value=10.0,
    description='YTM (%):'
)
coupon1_widget = widgets.FloatText(
    value=8.0,
    description='Coupon1 (%):'
)
coupon2_widget = widgets.FloatText(
    value=12.0,
    description='Coupon2 (%):'
)
coupon_freq_widget = widgets.Dropdown(
    options=[('Annual', 1), ('Semiannual', 2), ('Quarterly', 4)],
    value=1,
    description='Coupon frequency:'
)

output = widgets.Output()

def on_recalculate_clicked(_):
    with output:
        clear_output()
        plot_bond_price_dynamics(
            face_value_widget.value,
            ytm_widget.value,
            coupon1_widget.value,
            coupon2_widget.value,
            coupon_freq_widget.value
        )

recalculate_button = widgets.Button(description="Recalculate Graph")
recalculate_button.on_click(on_recalculate_clicked)

# Display one grouped interactive panel.
bond_controls = widgets.HBox([
    widgets.VBox([
        face_value_widget,
        ytm_widget,
        coupon_freq_widget,
    ]),
    widgets.VBox([
        coupon1_widget,
        coupon2_widget,
        recalculate_button,
    ]),
])
display(widgets.VBox([bond_controls, output]))

# Generate the default graph upon first execution.
on_recalculate_clicked(None)
```

### Yield to Call

*   If the bond is callable, it may be retired prior to the maturity date.
    *   At higher interest rates, the risk of call is negligible.
    *   At lower rates, the values of a callable and straight bond begin to diverge.
    *   The yield to call is calculated just like the yield to maturity
        *   The time until call replaces time until maturity,
        *   The call price replaces par value.


*   Sigma buys a 20-year bond for $\$1,120$. It has a $\$1,000$ par value and pays $\$40$ every six months. The bond can first be called in five years for $\$1,050$. Find its nominal annual yield to maturity and yield to first call.

$$ 1,120 = 40 \cdot \frac{1}{r/2} \left[1 - \frac{1}{(1+r/2)^{40}}\right] + \frac{1,000}{(1+r/2)^{40}} \quad \Rightarrow \quad r = \text{YTM} \approx 6.89\% $$

$$ 1,120 = 40 \cdot \frac{1}{r/2} \left[1 - \frac{1}{(1+r/2)^{10}}\right] + \frac{1,050}{(1+r/2)^{10}} \quad \Rightarrow \quad r = \text{Yield to first call} \approx 6.05\% $$

---

```python
# If needed, uncomment and run once:
# %pip install numpy matplotlib ipywidgets

import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, clear_output


def ytc_ytm_bond_value(nominal_yield, periods, coupon_payment,
                       redemption_value, payments_per_year):
    """Bond value for a nominal annual yield and a specified redemption date."""
    periodic_yield = nominal_yield / payments_per_year
    if periodic_yield <= -1:
        return np.nan
    if abs(periodic_yield) < 1e-12:
        return coupon_payment * periods + redemption_value

    discount_factors = (1 + periodic_yield) ** (-np.arange(1, periods + 1))
    return coupon_payment * discount_factors.sum() + redemption_value * discount_factors[-1]


def ytc_ytm_solve_yield(target_price, periods, coupon_payment,
                        redemption_value, payments_per_year):
    """Solve the nominal annual yield by bisection."""
    def price_gap(nominal_yield):
        return ytc_ytm_bond_value(
            nominal_yield, periods, coupon_payment,
            redemption_value, payments_per_year
        ) - target_price

    lower = -0.999 * payments_per_year
    upper = 1.0
    while price_gap(upper) > 0 and upper < 1_000:
        upper *= 2

    if price_gap(lower) < 0 or price_gap(upper) > 0:
        return np.nan

    for _ in range(120):
        midpoint = (lower + upper) / 2
        if price_gap(midpoint) > 0:
            lower = midpoint
        else:
            upper = midpoint
    return (lower + upper) / 2


def plot_ytc_ytm_convergence(face_value, coupon_rate_percent, maturity_years,
                             years_to_first_call, call_price, payments_per_year,
                             minimum_ytm_percent, maximum_ytm_percent,
                             reference_ytm_percent):
    if face_value <= 0 or call_price <= 0:
        print("Face value and call price must be positive.")
        return
    if coupon_rate_percent < 0:
        print("Coupon rate cannot be negative.")
        return
    if years_to_first_call <= 0 or maturity_years <= years_to_first_call:
        print("Years to first call must be positive and shorter than maturity.")
        return
    if minimum_ytm_percent >= maximum_ytm_percent:
        print("Minimum YTM must be below maximum YTM.")
        return
    if minimum_ytm_percent / 100 <= -payments_per_year:
        print("The selected minimum YTM is outside the valid discounting range.")
        return

    maturity_periods = maturity_years * payments_per_year
    call_periods = years_to_first_call * payments_per_year
    coupon_payment = face_value * (coupon_rate_percent / 100) / payments_per_year

    ytm_values = np.linspace(minimum_ytm_percent, maximum_ytm_percent, 301) / 100
    prices = np.array([
        ytc_ytm_bond_value(
            ytm, maturity_periods, coupon_payment,
            face_value, payments_per_year
        )
        for ytm in ytm_values
    ])
    ytc_values = np.array([
        ytc_ytm_solve_yield(
            price, call_periods, coupon_payment,
            call_price, payments_per_year
        )
        for price in prices
    ])

    valid = np.isfinite(ytc_values)
    if not valid.any():
        print("No valid YTC values could be calculated for the selected inputs.")
        return

    ytm_percent = ytm_values[valid] * 100
    ytc_percent = ytc_values[valid] * 100
    difference = ytc_percent - ytm_percent

    fig, ax = plt.subplots(figsize=(9, 5.5), dpi=110)
    ax.plot(ytm_percent, ytm_percent, linewidth=2.2, label="YTM")
    ax.plot(ytm_percent, ytc_percent, linewidth=2.2, label="Yield to First Call (YTC)")
    ax.fill_between(
        ytm_percent, ytm_percent, ytc_percent,
        color="slategray", alpha=0.15, label="Yield gap"
    )

    crossing_indices = np.where(difference[:-1] * difference[1:] <= 0)[0]
    if len(crossing_indices):
        index = crossing_indices[0]
        x0, x1 = ytm_percent[index], ytm_percent[index + 1]
        d0, d1 = difference[index], difference[index + 1]
        crossover = x0 if abs(d0) < 1e-12 else x0 - d0 * (x1 - x0) / (d1 - d0)
        crossover_price = ytc_ytm_bond_value(
            crossover / 100, maturity_periods, coupon_payment,
            face_value, payments_per_year
        )
        ax.scatter(crossover, crossover, color="black", marker="*", s=130, zorder=5)
        ax.axvline(crossover, color="black", linestyle=":", alpha=0.7)
        ax.annotate(
            f"Crossover: {crossover:.2f}%",
            (crossover, crossover), xytext=(8, -24),
            textcoords="offset points"
        )
        print(
            f"Yield to first call equals YTM at approximately {crossover:.2f}% "
            f"(bond price about ${crossover_price:,.2f})."
        )
    else:
        print("YTC and YTM do not cross within the selected YTM range.")

    reference_ytm = reference_ytm_percent / 100
    reference_price = ytc_ytm_bond_value(
        reference_ytm, maturity_periods, coupon_payment,
        face_value, payments_per_year
    )
    reference_ytc = ytc_ytm_solve_yield(
        reference_price, call_periods, coupon_payment,
        call_price, payments_per_year
    )
    if minimum_ytm_percent <= reference_ytm_percent <= maximum_ytm_percent:
        ax.scatter(
            reference_ytm_percent, reference_ytc * 100,
            color="crimson", s=55, zorder=5,
            label="Reference point"
        )
    print(
        f"At the reference YTM of {reference_ytm_percent:.2f}%, "
        f"price = ${reference_price:,.2f} and yield to first call = {reference_ytc * 100:.2f}%."
    )

    ax.set_title("Yield to First Call and Yield to Maturity")
    ax.set_xlabel("Assumed YTM (Nominal Annual Rate)")
    ax.set_ylabel("Nominal Annual Yield")
    ax.xaxis.set_major_formatter(plt.FuncFormatter(lambda x, _: f"{x:.1f}%"))
    ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda y, _: f"{y:.1f}%"))
    ax.grid(True, linestyle=":", alpha=0.5)
    ax.legend()
    plt.tight_layout()
    plt.show()


ytc_ytm_style = {"description_width": "145px"}
ytc_ytm_layout = widgets.Layout(width="310px")

ytc_ytm_face_widget = widgets.FloatText(
    value=1000, description="Face value:", style=ytc_ytm_style, layout=ytc_ytm_layout
)
ytc_ytm_coupon_widget = widgets.FloatText(
    value=8.0, description="Coupon rate (%):", style=ytc_ytm_style, layout=ytc_ytm_layout
)
ytc_ytm_maturity_widget = widgets.BoundedIntText(
    value=20, min=2, max=100, description="Maturity (years):",
    style=ytc_ytm_style, layout=ytc_ytm_layout
)
ytc_ytm_call_year_widget = widgets.BoundedIntText(
    value=5, min=1, max=99, description="Years to first call:",
    style=ytc_ytm_style, layout=ytc_ytm_layout
)
ytc_ytm_call_price_widget = widgets.FloatText(
    value=1050, description="Call price:", style=ytc_ytm_style, layout=ytc_ytm_layout
)
ytc_ytm_frequency_widget = widgets.Dropdown(
    options=[("Annual", 1), ("Semiannual", 2), ("Quarterly", 4)],
    value=2, description="Coupon frequency:",
    style=ytc_ytm_style, layout=ytc_ytm_layout
)
ytc_ytm_min_widget = widgets.FloatText(
    value=5.0, description="Minimum YTM (%):", style=ytc_ytm_style, layout=ytc_ytm_layout
)
ytc_ytm_max_widget = widgets.FloatText(
    value=7.5, description="Maximum YTM (%):", style=ytc_ytm_style, layout=ytc_ytm_layout
)
ytc_ytm_reference_widget = widgets.FloatText(
    value=6.886, description="Reference YTM (%):", style=ytc_ytm_style, layout=ytc_ytm_layout
)

ytc_ytm_output = widgets.Output()
ytc_ytm_button = widgets.Button(
    description="Recalculate Graph", button_style="info", icon="refresh"
)


def on_ytc_ytm_recalculate(_):
    with ytc_ytm_output:
        clear_output(wait=True)
        plot_ytc_ytm_convergence(
            ytc_ytm_face_widget.value,
            ytc_ytm_coupon_widget.value,
            ytc_ytm_maturity_widget.value,
            ytc_ytm_call_year_widget.value,
            ytc_ytm_call_price_widget.value,
            ytc_ytm_frequency_widget.value,
            ytc_ytm_min_widget.value,
            ytc_ytm_max_widget.value,
            ytc_ytm_reference_widget.value,
        )


ytc_ytm_button.on_click(on_ytc_ytm_recalculate)

display(widgets.VBox([
    widgets.HTML("<h3>Interactive Yield to First Call–YTM Comparison</h3>"),
    widgets.HBox([
        widgets.VBox([
            ytc_ytm_face_widget,
            ytc_ytm_coupon_widget,
            ytc_ytm_maturity_widget,
            ytc_ytm_call_year_widget,
            ytc_ytm_call_price_widget,
        ]),
        widgets.VBox([
            ytc_ytm_frequency_widget,
            ytc_ytm_min_widget,
            ytc_ytm_max_widget,
            ytc_ytm_reference_widget,
            ytc_ytm_button,
        ]),
    ]),
    ytc_ytm_output,
]))

on_ytc_ytm_recalculate(None)
```

- The annual market discount rate is 7%, with semiannual compounding.
- Both bonds can be called immediately for \$1,040. Assume each is called if its straight-bond value exceeds the call price.

### Bond Prices and Yields

Alpha pays $\$25$ and Beta pays $\$35$ every six months. Each has 12 semiannual periods remaining. The periodic discount rate falls from 3.5% to 2.5%.

$$
P_{\alpha,0} = 25 \cdot \frac{1-(1.035)^{-12}}{0.035} + \frac{1,000}{(1.035)^{12}} = \$903.37
$$

Beta's coupon rate matches the initial market rate, so it trades at par.
$$
P_{\beta,0} = \$1,000.00
$$

After the market rate falls to 5%, apply the $\$1,040$ call-price cap to each straight-bond value.

$$
P_{\alpha,1} = \min\left(25 \cdot \frac{1-(1.025)^{-12}}{0.025} + \frac{1,000}{(1.025)^{12}},\ 1{,}040\right)
$$

$$
P_{\alpha,1} = \min\left(1{,}000.00,\ 1{,}040.00\right) = \$1,000.00
$$

$$
P_{\beta,1} = \min\left(35 \cdot \frac{1-(1.025)^{-12}}{0.025} + \frac{1,000}{(1.025)^{12}},\ 1{,}040\right)
$$

$$
P_{\beta,1} = \min\left(1{,}102.58,\ 1{,}040.00\right) = \$1,040.00
$$

$$
\text{Capital gain (Alpha)} = \$1,000.00 - \$903.37 = \$96.63
$$

$$
\text{Capital gain (Beta)} = \$1,040.00 - \$1,000.00 = \$40.00
$$

#### Realized Compound Return versus Yield to Maturity

```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, HTML

# --- 1. Core Calculation Functions ---

def calculate_reinvestment_bond_price(face_value, coupon_rate, ytm, years, freq=1):
    periods = years * freq
    coupon_payment = (face_value * coupon_rate) / freq
    r = ytm / freq
    if r == 0:
        pv_coupons = coupon_payment * periods
    else:
        pv_coupons = coupon_payment * ((1 - (1 + r)**-periods) / r)
    pv_face_value = face_value / (1 + r)**periods
    return pv_coupons + pv_face_value

def calculate_realized_return(price, coupon_rate, years, reinvestment_rate):
    face_value = 1000.0
    coupon_payment = face_value * coupon_rate
    n_periods = years
    r = reinvestment_rate
    if r == 0:
        fv_of_all_coupons = coupon_payment * n_periods
    else:
        fv_of_all_coupons = coupon_payment * (((1 + r)**n_periods - 1) / r)
    total_future_value = fv_of_all_coupons + face_value
    if price > 0:
        realized_return = (total_future_value / price)**(1 / years) - 1
    else:
        realized_return = 0
    return total_future_value, realized_return

# --- 2. Plotting and Display Function ---

def generate_output(coupon, maturity, ytm, current_reinvest_rate):
    coupon_r = coupon / 100
    ytm_r = ytm / 100
    reinvest_r = current_reinvest_rate / 100
    
    initial_price = calculate_reinvestment_bond_price(1000.0, coupon_r, ytm_r, maturity)
    reinvest_rates_range = np.linspace(0, 0.15, 151)
    realized_returns_curve = [calculate_realized_return(initial_price, coupon_r, maturity, r)[1] for r in reinvest_rates_range]

    fv_at_point, realized_r_at_point = calculate_realized_return(initial_price, coupon_r, maturity, reinvest_r)

    color = "green" if np.isclose(realized_r_at_point, ytm_r, atol=1e-4) else ("red" if realized_r_at_point < ytm_r else "blue")
    summary_html = f"""
    <div style="font-family: Arial, sans-serif; font-size: 1.1em; line-height: 1.6;">
        <b>Initial Bond Purchase Price (V₀):</b> ${initial_price:,.2f}<br>
        <b>Promised Yield to Maturity (YTM):</b> {ytm:.2f}%<br>
        <hr>
        <b>Total Future Value at Maturity (Vₜ):</b> ${fv_at_point:,.2f} (at current reinvestment rate)<br>
        <hr>
        <b style="font-size: 1.2em;">Realized Compound Return: <span style="color:{color};">{realized_r_at_point*100:.2f}%</span></b>
    </div>
    """
    display(HTML(summary_html))

    fig, ax = plt.subplots(figsize=(10, 6))
    
    ax.plot(reinvest_rates_range * 100, np.array(realized_returns_curve) * 100, color='dodgerblue', linewidth=2.5, label='Realized Return')
    ax.axhline(y=ytm, color='grey', linestyle='--', label=f'Promised YTM ({ytm:.1f}%)')
    ax.plot(current_reinvest_rate, realized_r_at_point * 100, 'o', color='red', markersize=10, zorder=10)
    ax.text(current_reinvest_rate + 0.3, realized_r_at_point * 100, 
            f'({current_reinvest_rate:.1f}%, {realized_r_at_point*100:.2f}%)', 
            fontsize=11, verticalalignment='bottom')

    ax.set_title(f'Realized Return vs. Reinvestment Rate ({coupon}% Coupon, {maturity}-Year Bond)', fontsize=15)
    ax.set_xlabel('Reinvestment Rate (%)', fontsize=12)
    ax.set_ylabel('Realized Compound Return (%)', fontsize=12)
    ax.set_xlim(0, 15)
    ax.set_ylim(min(realized_returns_curve)*100 - 1, max(realized_returns_curve)*100 + 1)
    ax.legend()
    ax.grid(True, linestyle=':', alpha=0.7)
    
    plt.show()

# --- 3. Interactive Widgets Setup ---

style = {'description_width': 'initial'}
# --- FIX: Set continuous_update=True for all sliders to make them fully interactive ---
coupon_widget = widgets.FloatSlider(value=10.0, min=2, max=15, step=0.5, description='Coupon Rate (%):', style=style, continuous_update=True)
maturity_widget = widgets.IntSlider(value=2, min=2, max=30, step=1, description='Maturity (Years):', style=style, continuous_update=True)
purchase_ytm_widget = widgets.FloatSlider(value=10.0, min=2, max=15, step=0.5, description='Purchase YTM (%):', style=style, continuous_update=True)
reinvest_rate_widget = widgets.FloatSlider(value=10.0, min=0, max=15, step=0.1, description='Reinvestment Rate (%):', style=style, continuous_update=True)

# Capture each update in this demonstration's own output widget.
reinvestment_output = widgets.interactive_output(
    generate_output,
    {'coupon': coupon_widget, 'maturity': maturity_widget, 'ytm': purchase_ytm_widget, 'current_reinvest_rate': reinvest_rate_widget}
)

reinvestment_controls = widgets.VBox([
    widgets.HTML("<h3>1. Set Bond & Purchase Parameters:</h3>"),
    coupon_widget,
    maturity_widget,
    purchase_ytm_widget,
    widgets.HTML("<hr><h3>2. Explore with Reinvestment Rate Slider:</h3>"),
    reinvest_rate_widget
])

display(reinvestment_controls, reinvestment_output)
```

### The Term Structure of Interest Rates

Today's spot rates imply the following one-year forward rates:

$$ f_n = \frac{(1+y_n)^n}{(1+y_{n-1})^{n-1}} - 1, \qquad n \geq 2 $$

| Year $n$ | Spot rate $y_n$ | One-year forward rate $f_n$ |
| :---: | :---: | :---: |
| 1 | 5.00% | 5.00% |
| 2 | 6.00% | 7.01% |
| 3 | 7.00% | 9.03% |
| 4 | 8.00% | 11.06% |

*   $f_1 = y_1$ is today's one-year spot rate. Later forward rates are implied today; the actual future short rates are unknown.

### The Term Structure of Interest Rates

#### Spot rate
*   The yield to maturity on zero-coupon bonds.
*   The rate that prevails today for a time period corresponding to the zero's maturity.


#### Short rate
*   The one-period interest rate that prevails at a particular date. Future short rates are uncertain today.



#### Forward rate
*   The one-year rate implied by today's spot rates that makes an $n$-year zero-coupon investment equivalent to an $(n - 1)$-year zero-coupon investment rolled over for one year at that rate.
*   It equals the expected future short rate under the expectations hypothesis. A risk premium can make the two differ.
    *   If short-term investors dominate, the forward rate tends to exceed the expected future short rate to compensate for long-term bond price risk.
    *   If long-term investors dominate, the forward rate can be below the expected future short rate because rolling over short-term bonds involves reinvestment risk.


---

### The Term Structure of Interest Rates
#### Liquidity
* Liquidity refers to the ability to sell an asset easily at a predictable price.
* Long-term bonds have greater price risk, considered less liquid, must offer a premium.
* Liquidity Premium
    * The liquidity premium compensates short-term investors for the uncertainty about the price at which they will be able to sell their long-term bonds at the end of the year.

#### The Expectations Hypothesis
*   The forward rate equals the market consensus expectation of the future short interest rate,
*   $f_n = E(r_n)$, and liquidity premiums are zero.
*   Suggests that the yields on long-term bonds depend only on expectations of future short rates.
    *   The forward rates derived from the yield curve can be used to infer market expectations of future short rates.\

#### Liquidity Preference Theory
*   Short-term investors will be unwilling to hold long-term bonds unless the forward rate exceeds the expected short interest rate.
*   Long-term investors will be unwilling to hold short bonds unless the expected short interest rate exceeds the forward rate.
*   The liquidity preference theory of the term structure argues that short-term investors dominate the market so that the forward rate will generally exceed the expected short rate.

$$ f_n = E(r_n) + \text{Liquidity Premium} $$

### Question 2.1

Kappa observes a 1-year zero-coupon yield of 4% and a 2-year zero-coupon yield of 5%, both with annual compounding.

-   What is the implied 1-year forward rate for year 2? 

-   Under the Expectations Hypothesis, what does it imply about the expected short rate? 

-   Under Liquidity Preference Theory, could the expected short rate instead remain at 4%? Explain.

---

### Question 2.1: Answer

$$ f_{1,2} = \frac{(1.05)^2}{1.04} - 1 \approx 6.01\% $$

*   Under the Expectations Hypothesis, the liquidity premium is zero, so the expected 1-year rate in year 2 is about 6.01%.

*   Under Liquidity Preference Theory, the forward rate includes a positive liquidity premium. An expected rate of 4% is consistent with a premium of about 2.01 percentage points.

*   An upward-sloping yield curve therefore does not, by itself, show that short rates are expected to rise.
